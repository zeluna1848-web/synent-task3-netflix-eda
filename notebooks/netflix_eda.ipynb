```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
plt.style.use("ggplot")
```

```python
df = pd.read_csv("../data/netflix_titles.csv")
```

```python
# Create a copy
df_clean = df.copy()

# Fill missing values
df_clean["director"] = df_clean["director"].fillna("Unknown")
df_clean["cast"] = df_clean["cast"].fillna("Unknown")
df_clean["country"] = df_clean["country"].fillna("Unknown")

# Remove rows with missing values
df_clean = df_clean.dropna(subset=["date_added", "rating", "duration"])

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

print(df_clean.shape)
```

```python
df_clean[df_clean["rating"].str.contains("min", na=False)]
```

```python
import matplotlib.pyplot as plt

# Count Movies and TV Shows
type_counts = df_clean["type"].value_counts()

plt.figure(figsize=(6,5))
type_counts.plot(kind="bar")

plt.title("Movies vs TV Shows on Netflix")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.savefig("../images/chart_name.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(10,6))

df_clean["country"].value_counts().head(10).plot(kind="bar")

plt.title("Top 10 Countries with Most Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.savefig("../images/chart_name.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(10,6))

df_clean["rating"].value_counts().plot(kind="bar")

plt.title("Distribution of Netflix Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Count")

plt.xticks(rotation=45)

plt.savefig("../images/chart_name.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(12,6))

df_clean["release_year"].value_counts().sort_index().plot()

plt.title("Netflix Content Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.savefig("../images/chart_name.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(12,6))

df_clean["listed_in"].value_counts().head(10).plot(kind="barh")

plt.title("Top 10 Netflix Genres")
plt.xlabel("Number of Titles")

plt.savefig("../images/chart_name.png", dpi=300, bbox_inches="tight")
plt.show()
```

# Key Insights

1. Netflix has significantly more Movies than TV Shows.

2. The United States contributes the largest number of titles on Netflix.

3. TV-MA is the most common content rating, indicating that much of the content is intended for mature audiences.

4. Netflix's content library grew rapidly in recent years before stabilising.

5. Drama, International Movies, and Comedy are among the most common genres available on Netflix.

# Conclusion

The exploratory data analysis of the Netflix dataset provided valuable insights into the platform's content library.
